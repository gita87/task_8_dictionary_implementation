# Inspect DictFlow pipeline

Notebook sandbox untuk memeriksa hasil konversi DOCX ke CSV tanpa mengubah file sumber.

```python
from pathlib import Path
import pandas as pd
import sys

PROJECT_DIR = Path.cwd().resolve()
if PROJECT_DIR.name == 'sandbox':
    PROJECT_DIR = PROJECT_DIR.parent
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

from docx import Document
from qaos_dictionary import DEFAULT_IMAGE_QUALITY, extract_rows
from qaos_dictionary.core import _validated_docx_buffer

def convert_docx_to_dataframe(input_path, image_quality=DEFAULT_IMAGE_QUALITY):
    """Convert a dictionary DOCX directly to a DataFrame."""
    input_path = Path(input_path)
    with input_path.open('rb') as source:
        document = Document(_validated_docx_buffer(source))
    columns, rows = extract_rows(document, image_quality=image_quality)
    dataframe = pd.DataFrame(rows, columns=columns)
    dataframe['unique_id'] = dataframe['unique_id'].astype('string')
    return dataframe
```
