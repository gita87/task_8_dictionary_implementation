# Inspect DictFlow pipeline

Notebook sandbox untuk memeriksa hasil konversi DOCX ke CSV tanpa mengubah file sumber.

```python
from pathlib import Path
import csv
import io
import sys

PROJECT_DIR = Path.cwd().resolve()
if PROJECT_DIR.name == 'sandbox':
    PROJECT_DIR = PROJECT_DIR.parent
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

from dict_docx_to_csv import convert_dictionary_docx_to_csv
```

```python
# Ganti path ini untuk memeriksa DOCX lain.
input_path = PROJECT_DIR / 'tests' / 'fixtures' / 'golden' / 'production_dictionary.docx'
input_path.exists(), input_path
```

```python
with input_path.open('rb') as source:
    result = convert_dictionary_docx_to_csv(
        source,
        input_filename=input_path.name,
        image_quality=75,
    )

print({
    'filename': result.filename,
    'columns': result.columns,
    'row_count': result.row_count,
    'content_bytes': len(result.content),
})
```

```python
# Preview beberapa baris hasil tab-delimited CSV.
text = result.content.decode('utf-8-sig')
rows = list(csv.reader(io.StringIO(text), delimiter='\t'))
for row in rows[:6]:
    print(row)
```
